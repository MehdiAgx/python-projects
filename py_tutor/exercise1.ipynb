```python
def is_even(p):
    return (p%2 == 0)
print(is_even(61))
```

```python
def no_of_evens(nos):
    countt = 0
    for p in nos :
        if is_even(p):
            countt += 1
    return countt    
countt = no_of_evens([23, 67, 22, 66, 78, 44, 12, 34, 98, 38, 87, 72])
print(countt)            
```

```python
class circle():
    pi = 3.14
    def __init__(self, r):
        self.r = r 
    def area(self):
        a = self.r * self.r * circle.pi
        return a
        
```

```python
c = circle(34)
print(c.area())
```
